```python
#storing the password in a string variable
password = "pravallika@8"
#check if the password has atleast 8 characters
length = len(password) >= 8
#check if password contains "@" symbol
contains_sym = "@" in password
#check if both conditions are True
strong = length and contains_sym
#printing all three boolean results
print("Password:",password)
print("Length OK (>=8):",length)
print("Contains '@':",contains_sym)
print("Strong:",strong)
```

```python
#Storing 8 attendee names in a list
names=["Niha","Pravi","Kavya","Jai","Sai","Abhi","Veeha","Thridha"]
#Printing total no.of attendees
print("Total Attendees:", len(names))
#Printing the first attendee
print("First attendee:",names[0])
#Printing last 2 attendees using slicing
print("Last two attendees", names[-2:])
#sorting names alphabetically
sorting=sorted(names)
#Printing the names alphabetically
print("Names of the attendees alphabetically:", sorting) #print("first 3 alphabetically by directly",sorted(names)[:3])
#First 3 names alphabetically
print("First three attendees alphabetically", sorting[:3])
```

```python
#Creating a list of squares of even nums from 1 to 20
squares= [num ** 2 for num in range(1,21) if num % 2 == 0]
#Printing the resulting list
print("Squares of even numbers from 1 to 20", squares)
```

```python
#Creating a list of products along with their prices
products = [("Notebook",250),("Pen",20),("Backpack",1800),("Eraser",15),("Water Bottle",350)]
#Soring products by price using lambda function
sorted_products= sorted(products,key=lambda x:x[1])
#Printing the list
print("Sorted list", sorted_products)
```

```python
#Creating a dictionary with names and their scores
scores = {
"Anika": 35,
"Rohan": 72,
"Meera": 40,
"Kabir": 28,
"Diya": 88,
"Ishaan": 55,
}
#Filter students who scored 40 or more
passing=filter(lambda y: scores[y] >= 40,scores)
#Converting the result into a list
passing_students =  list(passing)
#Print the pass students
print("Passing students:", passing_students)
```
