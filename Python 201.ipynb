# **Track B - Text Article Analyzer**
The Text Article Analyzer is a Python program designed to analyze any article or text file and provides a clear summary of its structure.
The program prints a formatted summary containing:
*   Word count
*   Sentence count
*   Average word length
*   Estimated reading time
*   The most frequently used words







```python
# Function 1: Load the article
# load_article()
#Purpose: Opens article.txt and reads the article.
#It uses try/except so that if the file doesn't exist, the program shows an error instead of crashing.
def load_article(filename):
    try:
        with open(filename, "r") as file:
            text = file.read()
        return text

    except FileNotFoundError:
        print("Error: article.txt was not found.")
        return None
```

```python

# Function 2: Count words
#count_words()
#Purpose: Counts how many words are in the article.
def count_words(text):
    return len(text.split())
```

```python

# Function 3: Count sentences
#count_sentences()
#Purpose: Counts the number of sentences.
#It looks for periods (.) and counts the sentences that contain text
def count_sentences(text):
    sentences = text.split(".")
    count = 0

    for sentence in sentences:
        if sentence.strip():
            count += 1

    return count
```

```python
# Function 4: Calculate average word length
#average_word_length()
#Purpose: Calculates the average number of characters in a word.
#For example, if the words have lengths:
#cat = 3
#house = 5
#book = 4
#The average word length is:
#(3 + 5 + 4) / 3 = 4
def average_word_length(text):
    words = text.split()

    if len(words) == 0:
        return 0

    total_letters = 0

    for word in words:
        word = word.strip(".,!?;:'\"()[]{}")
        total_letters += len(word)

    return total_letters / len(words)
```

```python
# Function 5: Find frequently used words
#frequently_used_words()
#Purpose: Finds the top 5 most frequently used words in the article.
def frequently_used_words(text):
    words = text.lower().split()

    word_count = {}

    for word in words:
        word = word.strip(".,!?;:'\"()[]{}")

        if word:
            if word in word_count:
                word_count[word] += 1
            else:
                word_count[word] = 1

    # Sort words by frequency
    sorted_words = sorted(
        word_count.items(),
        key=lambda item: item[1],
        reverse=True
    )

    # Get only the top 5 words
    top_words = []

    for word, count in sorted_words[:5]:
        top_words.append(word)

    return top_words
```

```python

# Function 6: Calculate estimated reading time
#reading_time()
#Purpose: Estimates how long it takes to read the article.
#The program assumes an average reading speed of 200 words per minute
def reading_time(text):
    words = count_words(text)
    words_per_minute = 200

    return words / words_per_minute
```

```python
# Function 7: Display analysis
#display_analysis()
#Purpose: Displays all the results directly in the Colab output.
#It brings together:
#Word count
#Sentence count
#Average word length
#Frequently used words
#Estimated reading time
def display_analysis(text):
    print("\n========== ARTICLE ANALYSIS ==========")

    print("Word count          :", count_words(text))

    print("Sentence count      :", count_sentences(text))

    print("Average word length :",
          round(average_word_length(text), 2),
          "characters")

    # Frequently used words
    common_words = frequently_used_words(text)

    print("Frequently used words:", ", ".join(common_words))

    # Reading time
    time = reading_time(text)

    if time < 1:
        print("Estimated reading time: Less than 1 minute")
    else:
        print("Estimated reading time:",
              round(time, 1),
              "minutes")

    print("=======================================")
```

```python

# Main function
#main()
#Purpose: This is the main controlling function.
#It tells Python to:
#Load the article
#Check whether it exists
#Check whether it is empty
#Analyze it
#Display the results

def main():

    print("Starting Text Article Analyzer...")

    filename = "article.txt"

    # Load article
    article = load_article(filename)

    # Check if article was loaded
    if article is None:
        return

    # Check if article is empty
    if article.strip() == "":
        print("Error: article.txt is empty.")
        return

    # Display analysis directly
    display_analysis(article)

    print("Analysis completed successfully!")


# Start the program
main()
```
