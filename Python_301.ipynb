**1.Title: Clean, Analyse & Summarise a Real Dataset**

**Online Retail Data Analysis Report**

**2. Import Libraries**

```python
import pandas as pd
import numpy as np
```

**3. Load Dataset**

```python
url = "https://huggingface.co/datasets/aarav912/online-retail/resolve/main/online_retail.csv"
df = pd.read_csv(url)
```

**4. Understand the Data**

```python
#1. First few rows
print("--- Head ---")
display(df.head(10))
#summary Statistics
print("\n---summary Statistics (Min/Max values)---")
display(df.describe())
#2. Shape
print(f"\nShape: {df.shape[0]} rows, {df.shape[1]} columns")
#3. Column names
print("\nColumns:", df.columns.tolist())
#4. Data types
print("\n--- Data Types ---")
print(df.dtypes)
#5. Missing values
print("\n--- Missing Values ---")
print(df.isnull().sum())
#6. Duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())
```

**Data Understanding Summary:**
The 'Invoice data' column arrives as raw text rather than a datetime object and 'CustomerID' has a significant no.of missing values , and 'UnitPrice' along with 'Quantity' contain negative or zero values representing stock adjustments and order cancellations

**5. Clean the Data**

```python
# 1. Handle missing CustomerID
#Remove rows where CustomerID is missing
df = df.dropna(subset=['CustomerID'])
print("Shape after dropping missing CustomerID:", df.shape)

# 2. Convert InvoiceDate to datetime {From text(object) to proper datetime64 type}
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
print("Shape after datetime conversion:", df.shape)

# 3. Remove exact duplicate rows
#Identify and drop complete duplicate rows across all columns
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)

# 4. Remove invalid transactions (Quantity and UnitPrice must be positive)
#Filter out invalid entries
df = df[(df['Quantity'] > 0) & (df['UnitPrice'] > 0)]
print("Shape after filtering invalid transactions:", df.shape)
```

**6. Create Sales Column**

```python
# Calculate total sales per line item by multiplying Quantity by UnitPrice
df["Sales"] = df["Quantity"] * df["UnitPrice"]
# Verify updated shape and preview data with new Sales column
print("Shape after adding Sales column:", df.shape)
df.head()
```

**7. NumPy Analysis**

```python
# Extract the Sales column as a 1D NumPy array for array-based math
sales_data = df["Sales"].values

# Calculate statistical metrics using NumPy functions
mean_sales = np.mean(sales_data)  # Compute arithmetic mean
median_sales = np.median(sales_data)  # Compute 50th percentile (median)
min_sales = np.min(sales_data)  # Find the lowest sale value
max_sales = np.max(sales_data)  # Find the highest sale value
std_sales = np.std(sales_data)  # Compute standard deviation (spread)

# Print calculated values formatted to 2 decimal places
print(f"Mean Sales: {mean_sales:.2f}")
print(f"Median Sales: {median_sales:.2f}")
print(f"Minimum Sales: {min_sales:.2f}")
print(f"Maximum Sales: {max_sales:.2f}")
print(f"Standard Deviation: {std_sales:.2f}")
```

**8. Pandas Filtering**

```python
#1. Filter transactions with Sales value strictly greater than 500
sales_gt_500 = df[df["Sales"] > 500]
print(f"Transactions with Sales > 500: {len(sales_gt_500)}")

#2. Sort DataFrame by Sales descending and display the top 10 largest orders
print("\n--- Top 10 Largest Transactions ---")
display(df.sort_values(by="Sales", ascending=False).head(10))

#3. Filter rows where the Country column equals 'Germany'
germany_transactions = df[df["Country"] == "Germany"]
print(f"\nTransactions from Germany: {len(germany_transactions)}")

#4. Filter bulk transactions with item quantity greater than 1000
large_qty = df[df["Quantity"] > 1000]
print("\n--- Transactions with Quantity > 1000 ---")
display(large_qty)
```

**9. GroupBy & Aggregation**

```python
# Group data by Country and aggregate metrics (count, sum, mean)
country_summary = (
    df.groupby("Country")
    .agg(
        Number_of_Transactions=("InvoiceNo", "count"),  # Count total orders
        Total_Sales=("Sales", "sum"),  # Calculate cumulative revenue
        Average_Sales=("Sales", "mean"),  # Calculate average transaction value
    )
    .reset_index()
)  # Convert group keys back into regular DataFrame columns

# Sort countries by Total_Sales in descending order
country_summary_sorted = country_summary.sort_values(by="Total_Sales", ascending=False
)
display(country_summary_sorted)
```

**10. Merge / Region Analysis**

```python
# Define dictionary mapping country names to broad geographic regions
region_lookup = {
    "United Kingdom": "Europe",
    "France": "Europe",
    "Germany": "Europe",
    "Spain": "Europe",
    "Netherlands": "Europe",
    "Belgium": "Europe",
    "Switzerland": "Europe",
    "Portugal": "Europe",
    "Australia": "Oceania",
    "Japan": "Asia",
    "USA": "Americas",
}

# Convert python dictionary into a 2-column DataFrame
regions_df = pd.DataFrame(
    list(region_lookup.items()), columns=["Country", "Region"]
)

# Perform a left join to attach Region data to country summary records
merged_summary = pd.merge(country_summary, regions_df, on="Country", how="left")

# Fill unmapped/missing country regions with the default string 'Other'
merged_summary["Region"] = merged_summary["Region"].fillna("Other")

# Aggregate total sales grouped by broad geographic Region
region_summary = (
    merged_summary.groupby("Region")["Total_Sales"]
    .sum()
    .reset_index()
    .sort_values(by="Total_Sales", ascending=False)
    .reset_index(drop=True)
)
display(region_summary)
```

Which region generated the highest total sales?

Europe generated the highest total sales by a wide margin, driven primarily by the high overall transaction volume coming from the United Kingdom.
